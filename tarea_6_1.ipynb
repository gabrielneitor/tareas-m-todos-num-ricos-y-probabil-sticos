```python
#6.1
# impotamos numpy para funciones como seno ycoseno y matplotlib.pyplot para las garficas
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return (1 - np.cos(x) - (np.sin(x)/x) )# la funcion solisitada para que encontremos un 0 en 6 . 6,5

def biseccion(f, a, b, eps, N=1000000)
:#funcion de bisseccion vista en clases f[funcion a evaluar], a y b cota menor y superior respectivamente , eps un numero muy pequeño que nos sirve como filtro, N numero de veses que repetiremos el proseso
    pp = a
    for i in range(N):
        p = a + (b - a)/2

        if abs(p - pp)/abs(p) < eps: #filtro para saver si ya se encontro el 0 o no
            return p

        elif f(p)*f(a) > 0: #encaso que antes no se aya encontrado un 0 cambiamos a o b a dependiendo del valor del producto f(p)*f(a)
            a = p
        else:
            b = p
    pp = p
    print("No se encontró el cero después de", N, "iteraciones.")# en caso de que no ayamos encontrado el 0 en N iteraciones

print(biseccion(f, 6.0, 6.5, 10**(-10))) #nos muestra si hay un 0 en el intervalo seleccionado

point = 50

def error(N, a, b):#funcion error mostrada en clases
    return 2**(-N) * (b-a) # aqui tube problemas al no saver como interpretar N ya que porlo que entendi N aqui es la misma N que en bisseccion pero al ser ese N muy grande y al estar como exponente desidi evaluar con un numero menos

iter = range(1, point + 1) # cantidad de puntos
errorpoint = [error(n, 6.0, 6.5) for n in iter] #cantidad de puntos en la grafica

plt.loglog(iter, errorpoint, label='Cota del error de bisección')
plt.xlabel('Número de iteraciones')
plt.ylabel('Cota del error')
plt.legend()
plt.grid(True)
plt.show()
```
