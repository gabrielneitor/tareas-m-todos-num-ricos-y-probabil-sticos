```python
#9.1
import numpy as np
#para esta tarea tomare en supocicion que a lo que usted se refiere con evaluaciones es al numero de ecuaciones dentro de rk4 y rkf que contienen a f(t,y)

y0 = 0#valor inicial dado en la tarea
a = 0#punto de inicio del intervalo
b = 1#punto final del intervalo
tol = 10**-6#toleracia pedida en la tarea

def f(t, y):#derivada de la funcion que buscamos que depende de t e y
    return (t*np.exp(3*t) - 2*y)

def y(t):
    return((1/5)*t*np.exp(3*t) - (1/25)*np.exp(3*t) + (1/25)*np.exp(-2*t))

def rk4(f, a, b, yo, N):#funcion tomada directamente de la clase no profundisare mucho en ello solo mensionare los cambios

    h = (b-a)/N
    t = a
    w = yo
    ts = np.zeros(N+1)
    sol = np.zeros(N+1)
    evalk4 = 0

    for i in range(N):
        ts[i] = t
        sol[i] = w
        k1 = h*f(t, w)
        k2 = h*f(t+h/2, w+k1/2)
        k3 = h*f(t+h/2, w+k2/2)
        k4 = h*f(t + h, w + k3)
        w = w + (k1 + 2*k2 + 2*k3 + k4)/6
        t = t + h
        evalk4 += 4 #como aqui hay 4 funciones que calculan f(t,y) solo le sumamos 4 en cada iteracion de la funcion for


    ts[N] = t
    sol[N] = w
    print("evaluaciones totales", evalk4)
    print("=====================================================================")
    return ts, sol , evalk4
# ===============================================separacion para no perderme=========================================================
class MinStep(Exception):
    pass

def rkf(f, a, b, yo, tol, hmin, hmax):#funcion dada en clases no profindisare en ella solo los cambios echos

    t = a
    w = yo
    h = hmax
    flag = 1

    ts = [t]
    sol = [w]
    hs = []
    rs = []
    evalkf = 0 #valor para tomar rejistro de el numero de evaluaciones

    while flag:

        # Evaluaciones de la función
        k1 = h*f(t, w)
        k2 = h*f(t + h/4, w + k1/4)
        k3 = h*f(t + 3*h/8, w + 3*k1/32 + 9*k2/32)
        k4 = h*f(t + 12*h/13, w + 1932*k1/2197 - 7200*k2/2197 + 7296*k3/2197)
        k5 = h*f(t + h, w + 439*k1/216 - 8*k2 + 3680*k3/513 - 845*k4/4104)
        k6 = h*f(t + h/2, w - 8*k1/27 + 2*k2 - 3544*k3/2565 + 1859*k4/4104 - 11*k5/40)
        evalkf += 6 #como aqui hay 6 funciones que calculan f(t,y) solo le sumamos 6 en cada iteracion de la funcion while

        R = abs(k1/360 - 128*k3/4275 - 2197*k4/75240 + k5/50 + 2*k6/55)

        if R <= tol:
            t += h
            w = w + 25*k1/216 + 1408*k3/2565 + 2197*k4/4104 - k5/5
            ts.append(t)
            sol.append(w)
            hs.append(h)
            rs.append(R)

        q = 0.84*(tol/R)**(1/4)

        if q <= 0.1:
            h = 0.1*h
        elif q >= 4:
            h = 4*h
        else:
            h = q*h
        if h > hmax:
            h = hmax

        if t >= b:
            flag = 0
        elif t + h > b:
            h = b - t
        elif h < hmin:
            raise MinStep('Se excedió el paso mínimo.')
    print("evaluaciones totales", evalkf)
    print("=====================================================================")
    return np.array(ts), np.array(sol), np.array(hs), np.array(rs), evalkf

tsrk4, solrk4, eval_rk4 = rk4(f, a, b, y0, N=60)#los valores de N hmin y hmax fueron puesto por prueba y error
tsrkf, solrkf, _, _, eval_rkf = rkf(f, a, b, y0, tol, hmin=10**-7, hmax=20**-1)

err_rk4 = np.max(np.abs(y(tsrk4) - solrk4))#ccalculo las funciones y(t) con los ts dados tanto por los metodos rkf y rk4 para poder conparar la solucion con los metos aqui presente
err_rkf = np.max(np.abs(y(tsrkf) - solrkf))#np.max es para no tener un listados de valores sino tomo el valor mas grande para compararlos porque estos marcan el maximo error echo por la pc

print("Error máximo rk4 =", err_rk4,"======= evaluaciones =", eval_rk4)
print("Error máximo rkf =", err_rkf,"======= evaluaciones =", eval_rkf)

#conclucion el metodo de rk4 tiene el mismo orden de precicion de el metodo rkf pero el metodo de rkf cumple con el marjen de error en muchos menos pasos eso tomando en cuenta
#el pequeño valor de hmax que propuse por ende creo que el metodo de rkf es mejor llega a errores del mismo horden que rk4 pero en la mitad de paso
```

```python
10.1
import numpy as np
import matplotlib.pyplot as plt
#valores dados en la tarea
tet1 = np.array([0.1, 0])
tet2 = np.array([1, 0])
tet3 = np.array([3, 0])
g = 9.81
L = 1
b = 6
a = 0
N = 300 #cantidad de

def rk4s(f, a, b, yo, N):

    h = (b-a)/N
    t = a
    w = yo
    m = len(yo)
    ts = np.zeros(N+1)
    sol = np.zeros((m,N+1))

    for i in range(N):
        ts[i] = t
        sol[:,i] = w
        k1 = h*f(t, w)
        k2 = h*f(t+h/2, w+k1/2)
        k3 = h*f(t+h/2, w+k2/2)
        k4 = h*f(t + h, w + k3)
        w = w + (k1 + 2*k2 + 2*k3 + k4)/6
        t = t + h

    ts[N] = t
    sol[:,N] = w

    return ts, sol

def f(t, y): #funcion exsacta dada en la tarea
    return np.array([y[1], -(g / L) * np.sin(y[0])])
def fapp(t, y): #funcion aproximada con pequeñas variaciones de teta
    return np.array([y[1], -(g / L) * y[0]])
#graficos con cada valor de teta
#1
ts1 ,sol1 = rk4s(f, a, b, tet1, N)
tsapp1, solapp1 = rk4s(fapp, a, b, tet1, N)

plt.plot(ts1, sol1[0], label='Exacta ($\\sin\\theta$)')
plt.plot(tsapp1, solapp1[0], '--', label='Pequeñas oscilaciones ($\\theta$)')
plt.title('Comparación para $\\theta(0) = 0.1$')
plt.xlabel('t (s)')
plt.ylabel('$\\theta(t)$ (rad)')
plt.legend()
plt.grid(True)
plt.show()

#2
ts2 ,sol2 = rk4s(f, a, b, tet2, N)
tsapp2, solapp2 = rk4s(fapp, a, b, tet2, N)

plt.plot(ts2, sol2[0], label='Exacta ($\\sin\\theta$)')
plt.plot(tsapp2, solapp2[0], '--', label='Pequeñas oscilaciones ($\\theta$)')
plt.title('Comparación para $\\theta(0) = 1$')
plt.xlabel('t (s)')
plt.ylabel('$\\theta(t)$ (rad)')
plt.legend()
plt.grid(True)
plt.show()

#3
ts3 ,sol3 = rk4s(f, a, b, tet3, N)
tsapp3, solapp3 = rk4s(fapp, a, b, tet3, N)

plt.plot(ts3, sol3[0], label='Exacta ($\\sin\\theta$)')
plt.plot(tsapp3, solapp3[0], '--', label='Pequeñas oscilaciones ($\\theta$)')
plt.title('Comparación para $\\theta(0) = 3$')
plt.xlabel('t (s)')
plt.ylabel('$\\theta(t)$ (rad)')
plt.legend()
plt.grid(True)
plt.show()

#en la primera grafica la aproximacion se mantiene muy presisa a la exsacta
#segunda grafica la aproximacion tiene la misma amplitud pero esta desfasada de la exsacta
#tercera la aproximacion pierde toda presicion tiene un desfase enorme y la amplitud es varias veses mas pequeño que la exsacta
```
