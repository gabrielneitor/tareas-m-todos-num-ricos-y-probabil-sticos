```python
#8.8
import numpy as np
import matplotlib.pyplot as plt

def Y(t): #apartir de ecuaciones diferenciales pude encontrar esta ecuacion como sulucion
  return(np.arccos(np.sin(t)-np.cos(1)))

def dY(t,y): #primera derivada dada en la tarea
  return(np.cos(t)/np.sin(y))

def d2Y(t,y): #segunda derivada sacada a partir de regla de la cadena tambien la tercera
     return (-np.sin(t) / np.sin(y) - (np.cos(t)**2 * np.cos(y)) / np.sin(y)**3)

def d3Y(t, y): #tercera derivada
    return (-np.cos(t) / np.sin(y) + (3 * np.sin(t) * np.cos(t) * np.cos(y)) / np.sin(y)**3 + np.cos(t)**3 / np.sin(y)**3 + (3 * np.cos(t)**3 * np.cos(y)**2) / np.sin(y)**5)

y0 = 1

def meteuler(f, a, b, yo, N):#metodo euler

    h = (b-a)/N
    t = a
    w = yo
    ts = np.zeros(N+1)
    sol = np.zeros(N+1)

    for i in range(N):
        ts[i] = t
        sol[i] = w
        w = w + h*f(t, w)
        t = t + h

    ts[N] = t
    sol[N] = w

    return ts, sol

def taylor2(f, df, a, b, yo, N): #metodo teylor orden 2

    h = (b-a)/N
    t = a
    w = yo
    ts = np.zeros(N+1)
    sol = np.zeros(N+1)

    for i in range(N):
        ts[i] = t
        sol[i] = w
        w = w + h*f(t, w) + h**2*df(t, w)/2
        t = t + h

    ts[N] = t
    sol[N] = w

    return ts, sol

def taylor3(f, df, d2f, a, b, yo, N): #metodo teylor orden 3

    h = (b-a)/N
    t = a
    w = yo
    ts = np.zeros(N+1)
    sol = np.zeros(N+1)

    for i in range(N):
        ts[i] = t
        sol[i] = w
        w = w + h*f(t, w) + h**2*df(t, w)/2 + h**3*d2f(t, w)/6
        t = t + h

    ts[N] = t
    sol[N] = w

    return ts, sol

N = 10 #intervalo donde veremos la aproximacion
#evaluamos cada metodo para calcular su error
tseuler, soleuler = meteuler(dY, 0, 1, y0, N)

tstaylor2, soltaylor2 = taylor2(dY, d2Y, 0, 1, y0, N)

tstaylor3, soltaylor3 = taylor3(dY, d2Y, d3Y, 0, 1, y0, N)

# Calculamos la solucion exacta para la comparacion
exacta = np.arccos(np.cos(1) - np.sin(tseuler))

# evaluacion de errores
ereuler = exacta - soleuler
ertaylor2 = exacta - soltaylor2
ertaylor3 = exacta - soltaylor3

for i in range(N+1):
    print(f"{tseuler[i]:.2f} "
          f"{ereuler[i]:.10f} "
          f"{ertaylor2[i]:.10f} "
          f"{ertaylor3[i]:.10f} ")


# Graficamos los errores
plt.plot(tseuler, ereuler, "o-", label="Euler")
plt.plot(tstaylor2, ertaylor2, "o-", label="Taylor 2")
plt.plot(tstaylor3, ertaylor3, 'o-', label="Taylor 3")

plt.xlabel("t")
plt.ylabel("Error")
plt.legend()
plt.grid()
plt.show()


  #lo que espero que suseda en este caso es que el error de euler cresca mas mientras mas grande sea el intervalo de evalucaio
  # y que el error de euler va a ser mas grande que que las de teylor y que el error de teylor 2 sera mayor que la de orden 3 por ser de un grado mayor
```
